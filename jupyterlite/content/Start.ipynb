# Python-Programme

Die drei Notebooks ergänzen den Beitrag
*Was hat ein Platz, was ein anderer nicht hat?* um Simulationen und
empirische Untersuchungen.

Die veränderbaren Eingaben stehen jeweils zu Beginn. Mit `seed = 42`
sind die Simulationen reproduzierbar.

## Auswahl

1. [Wilson-Konfidenzintervall - Ellipse](Konfidenzellipse.ipynb)  
   Konfidenzellipse mit Prognoseintervallen.

2. [Wilson-Konfidenzintervall - Ellipse](Konfidenzellipse.ipynb)  
   Konfidenzellipse mit Prognoseintervallen.

3. [Verteilung H=X/n](H-Binom.ipynb)  
   Die diskrete Verteilung $H=X/n$; $X \sim \text{Bin}(n,p).$

4. [Nahaufnahme - Konfidenzellipe](Nahaufnahme-ph-Raum.ipynb)  
   Nahaufnahme der Konfidenzellipse (Lösungshinweis für eine Aufgabe aus dem Zentralabitur 2025).

5. [Verfahrenswahrscheinlichkeit](Verfahrenswkeit-p0.ipynb)  
   Verfahrenswahrscheinlichkeit bei festem $p_0$. Die linke Grafik zeigt Realisationen von  auf der Vertikalen . Die rechte Grafik zeigt die zugehörigen Wilson-Konfidenzintervalle.
